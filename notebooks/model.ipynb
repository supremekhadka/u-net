```python
import torch
from unet import UNet
```

```python
DEVICE = torch.device("cuda:0") if torch.cuda.is_available() else torch.device("cpu")

model = UNet(in_channels=3, out_channels=10)
model = model.to(DEVICE)

input = torch.rand((1, 3, 572, 572), device=DEVICE)

with torch.no_grad():
    output = model(input)

model.to("cpu")

output.cpu().detach()

output.shape
```
