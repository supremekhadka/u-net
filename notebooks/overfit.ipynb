```python
import torch
from unet import run_training, BriscDataset, BriscTransform
from unet.train import load_model
from torch.utils.data import DataLoader, Subset
```

```python
train_transforms = BriscTransform(
        size=572,
        mean=0.195,
        std=0.168,
        train=True
    )

val_transforms = BriscTransform(
        size=572,
        mean=0.195,
        std=0.168,
        train=False
    )
```

```python
train_dataset = BriscDataset(
        images_root="../../data/brisc2025",
        csv="../../data/brisc2025/metadata.csv",
        split="train",
        transforms=train_transforms
    )

val_dataset = BriscDataset(
        images_root="../../data/brisc2025",
        csv="../../data/brisc2025/metadata.csv",
        split="val",
        transforms=val_transforms
    )

len(train_dataset), len(val_dataset)
```

```python
train_subset = Subset(train_dataset, [0,1])
val_subset = Subset(train_dataset, [0,1])
```

```python
train_dataloader = DataLoader(
        dataset=train_subset, 
        batch_size=2
    )

val_dataloader = DataLoader(
        val_subset, 
        batch_size=2
    )
```

```python
DEVICE = torch.device("cuda:0") if torch.cuda.is_available() else torch.device("cpu")

model = load_model(
    device=DEVICE,
    in_channels=1,
    out_channels=4,
    base_channels=64,
    depth=4
)

print(model)
```

```python
run_training(
    model=model,
    device=DEVICE,
    train_dataloader=train_dataloader,
    val_dataloader=val_dataloader,
    criterion=torch.nn.CrossEntropyLoss(),
    optimizer=torch.optim.AdamW(model.parameters(), lr=1e-3),
    epochs=500,
    save_dir="../weights/overfit"
)
```
